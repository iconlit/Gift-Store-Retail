**THIS IS THE 2009-2010 CLEANING**

**import all libraries**

```python
import pandas as pd
import numpy as np
```

**LOAD THE CSV**

```python
df = pd.read_excel("../data/online_retail_II.xlsx",sheet_name="Year 2009-2010")
```

**display the csv**

```python
df.head()
```

**csv info**

```python
df.info()
```

**columns in list**

```python
object_cols = ["Invoice","StockCode","Description"]
```

**fixing df types**

```python
df[object_cols] = df[object_cols].astype("string")
```

**turning column names to lowercase and removing spacing**

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","",regex=False)
```

**turning column values to lower case**

```python
col_str = ["description","country","stockcode"]

df[col_str] = df[col_str].apply(lambda x:x.str.lower())
df[col_str] = df[col_str].apply(lambda x:x.str.strip())
```

**working on the time column seperating them into its own column**

```python
df["year"] = df["invoicedate"].dt.year
df["month"] = df["invoicedate"].dt.month_name()
df["day"] = df["invoicedate"].dt.day_name()
# df["time"] = df["invoicedate"].dt.time

df["hour"] = df["invoicedate"].dt.hour
```

**adding new columns**

```python
df["returned"] = np.where(df["quantity"] < 0,"yes","no")

df["gross_sales"] = np.where(df["quantity"] < 0,0,
                             df["quantity"] * df["price"])

df["net_sales"] = df["quantity"] * df["price"]
```
